# Retrieve-for-Train — Simplified Pseudocode

Conceptual notebook using the **Home Office Setup** example.

> Large model learns good retrieval behavior offline; a small diffusion retriever executes it efficiently online.

```python
# PHASE 1: OFFLINE TRAINING
for query in training_queries:
    subqueries = large_llm.generate_fanout(query, num_queries=10)
    alignment = measure_alignment(query, subqueries)
    diversity = measure_diversity(subqueries)
    groundedness = measure_groundedness(subqueries, product_database)
    reward = alignment + diversity + groundedness
    update_large_llm(subqueries, reward)
```

```python
# PHASE 2: SUPERVISION SYNTHESIS
training_examples = []
for query in many_queries:
    good_search_set = trained_large_llm.generate_fanout(query)
    training_examples.append((query, good_search_set))

# Example:
# Home office setup -> chair, desk, monitor, keyboard, lamp, webcam, headset
```

```python
# PHASE 3: TRAIN SMALL DIFFUSION RETRIEVER
for query, target_set in training_examples:
    query_embedding = embed(query)
    target_embeddings = embed(target_set)
    diffusion_retriever.train(query_embedding, target_embeddings)
```

```python
# ONLINE / INFERENCE
def search(user_query):
    query_embedding = embed(user_query)
    retrieval_embeddings = diffusion_retriever.generate(query_embedding)
    return vector_database.search(retrieval_embeddings)
```

## Architecture

**Traditional:** User Query → Large LLM → Reason → Subqueries → Retrieval

**Retrieve-for-Train:** Large LLM + RL (offline) → supervision → small diffusion retriever → fast online retrieval

### Key takeaway
**Training time = Learn the intelligence 🧠**  
**Inference time = Execute it efficiently ⚡**
